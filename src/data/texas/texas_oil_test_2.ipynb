```python
import pandas as pd
import geopandas as gpd
import numpy as np
import folium
from folium.plugins import HeatMap
```

```python
# 1. File path and column names
# -----------------------------

parquet_path = "../../../data/raw/texas/coordinates/production_with_lease_coordinates.parquet"

LAT_COL = "lease_latitude"        # change if needed
LON_COL = "lease_longitude"       # change if needed
FLARE_COL = "lease_csgd_tot_disp"
```

```python
# 2. Load parquet file
# -----------------------------
df = pd.read_parquet(parquet_path)

print(df.head())
print(df.columns)
```

```python
# 3. Basic cleaning
# -----------------------------
df = df[[LAT_COL, LON_COL, FLARE_COL]].copy()

df[LAT_COL] = pd.to_numeric(df[LAT_COL], errors="coerce")
df[LON_COL] = pd.to_numeric(df[LON_COL], errors="coerce")
df[FLARE_COL] = pd.to_numeric(df[FLARE_COL], errors="coerce")

df = df.dropna(subset=[LAT_COL, LON_COL, FLARE_COL])
```
